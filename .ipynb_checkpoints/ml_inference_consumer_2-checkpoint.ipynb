```python
from kafka import KafkaConsumer
import json
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import time
from collections import defaultdict
```

```python
# ── Kafka configuration ────────────────────────────────────────────────────────
OUTPUT_TOPIC            = "aeration_predictions"   # topic written by the ML producer
KAFKA_BOOTSTRAP_SERVERS = ["localhost:9092"]
GROUP_ID                = "dashboard_group"

MAX_POINTS = 100   # rolling window per sensor
PLOT_EVERY = 5     # regenerate PNGs every N messages
```

```python
sensor_store = defaultdict(lambda: {
    'all_timestamps':   [],
    'all_predictions':  [],
    'ts_aeration':      [],
    'conf_aeration':    [],
    'ts_no_aeration':   [],
    'conf_no_aeration': [],
})
msg_counter = 0
```

```python
def plot_dashboard(sensor_id, sd):
    """
    Save a 3-subplot dashboard PNG for one sensor.
    3.3.1 - aeration state (0/1) for every timestamp
    3.3.2 - proba_aeration when prediction == 1
    3.3.3 - proba_no_aeration when prediction == 0
    """
    fig, axes = plt.subplots(3, 1, figsize=(12, 10))
    fig.suptitle(f'Inference Dashboard - Sensor {sensor_id}', fontsize=13, fontweight='bold')

    # 3.3.1 - Aeration state over time
    ax1 = axes[0]
    if sd['all_timestamps']:
        ax1.step(sd['all_timestamps'], sd['all_predictions'],
                 where='post', color='royalblue', linewidth=1.5)
        ts_on  = [t for t, p in zip(sd['all_timestamps'], sd['all_predictions']) if p == 1]
        ts_off = [t for t, p in zip(sd['all_timestamps'], sd['all_predictions']) if p == 0]
        if ts_on:  ax1.scatter(ts_on,  [1]*len(ts_on),  color='green', s=40, zorder=5, label='Aeration ON')
        if ts_off: ax1.scatter(ts_off, [0]*len(ts_off), color='red',   s=40, zorder=5, label='Aeration OFF')
    ax1.set_yticks([0, 1])
    ax1.set_yticklabels(['No aeration (0)', 'Aeration (1)'])
    ax1.set_title('3.3.1 - Aeration state per timestamp')
    ax1.set_ylabel('Prediction')
    ax1.set_xlabel('Timestamp (epoch s)')
    ax1.legend(loc='upper right', fontsize=8)
    ax1.tick_params(axis='x', rotation=20)

    # 3.3.2 - proba_aeration when prediction == 1
    ax2 = axes[1]
    if sd['ts_aeration']:
        x = range(len(sd['ts_aeration']))
        ax2.bar(x, sd['conf_aeration'], color='mediumseagreen', width=0.6)
        ax2.set_xticks(list(x))
        ax2.set_xticklabels([str(t)[-5:] for t in sd['ts_aeration']],
                            rotation=45, ha='right', fontsize=7)
    else:
        ax2.text(0.5, 0.5, 'No aeration predictions yet',
                 ha='center', va='center', transform=ax2.transAxes, color='gray')
    ax2.set_ylim(0, 1.05)
    ax2.axhline(0.5, color='gray', linestyle='--', linewidth=0.8)
    ax2.set_title('3.3.2 - P(aeration) confidence when prediction = 1')
    ax2.set_ylabel('proba_aeration')
    ax2.set_xlabel('Timestamp (last 5 digits)')

    # 3.3.3 - proba_no_aeration when prediction == 0
    ax3 = axes[2]
    if sd['ts_no_aeration']:
        x = range(len(sd['ts_no_aeration']))
        ax3.bar(x, sd['conf_no_aeration'], color='tomato', width=0.6)
        ax3.set_xticks(list(x))
        ax3.set_xticklabels([str(t)[-5:] for t in sd['ts_no_aeration']],
                            rotation=45, ha='right', fontsize=7)
    else:
        ax3.text(0.5, 0.5, 'No no-aeration predictions yet',
                 ha='center', va='center', transform=ax3.transAxes, color='gray')
    ax3.set_ylim(0, 1.05)
    ax3.axhline(0.5, color='gray', linestyle='--', linewidth=0.8)
    ax3.set_title('3.3.3 - P(no_aeration) confidence when prediction = 0')
    ax3.set_ylabel('proba_no_aeration')
    ax3.set_xlabel('Timestamp (last 5 digits)')

    plt.tight_layout()
    fname = f'dashboard_sensor_{sensor_id}.png'
    fig.savefig(fname, dpi=100)
    plt.close(fig)
    print(f'  Dashboard saved -> {fname}')
```

```python
# ── Initialize consumer ─────────────────────────────────────────────────────────
consumer = KafkaConsumer(
    OUTPUT_TOPIC,
    bootstrap_servers  = KAFKA_BOOTSTRAP_SERVERS,
    group_id           = GROUP_ID,
    auto_offset_reset  = 'latest',
    enable_auto_commit = True,
    value_deserializer = lambda m: json.loads(m.decode('utf-8')),
    key_deserializer   = lambda k: k.decode('utf-8') if k else None,
)

print(f"Subscribed to Kafka topic '{OUTPUT_TOPIC}'.")
print('Waiting for inference results... (Kernel -> Interrupt to stop)\n')
```

```python
# ── Main loop ───────────────────────────────────────────────────────────────────
try:
    for message in consumer:
        result    = message.value
        sensor_id = result.get('sensor_id', message.key or 'unknown')
        ts        = result['timestamp']
        pred      = result['prediction']
        p_aer     = result['proba_aeration']
        p_no_aer  = result['proba_no_aeration']

        # Print to screen (same format as the ML producer)
        label = 'AERATION' if pred == 1 else 'no aeration'
        print(
            f"Sensor {sensor_id} | ts={ts} "
            f"| Pred: {pred} ({label}) "
            f"| P(0): {p_no_aer} "
            f"| P(1): {p_aer}"
        )

        # Store
        sd = sensor_store[sensor_id]
        sd['all_timestamps'].append(ts)
        sd['all_predictions'].append(pred)

        if pred == 1:      # aeration chosen → store aeration confidence
            sd['ts_aeration'].append(ts)
            sd['conf_aeration'].append(p_aer)
        else:              # no-aeration chosen → store no-aeration confidence
            sd['ts_no_aeration'].append(ts)
            sd['conf_no_aeration'].append(p_no_aer)

        # Rolling window
        for key in ('all_timestamps', 'all_predictions'):
            if len(sd[key]) > MAX_POINTS:
                sd[key] = sd[key][-MAX_POINTS:]
        for key in ('ts_aeration', 'conf_aeration', 'ts_no_aeration', 'conf_no_aeration'):
            if len(sd[key]) > MAX_POINTS:
                sd[key] = sd[key][-MAX_POINTS:]

        msg_counter += 1

        # Regenerate dashboards every PLOT_EVERY messages
        if msg_counter % PLOT_EVERY == 0:
            for sid, sd in sensor_store.items():
                plot_dashboard(sid, sd)

except KeyboardInterrupt:
    print('\nStopped. Generating final dashboards...')
    for sid, sd in sensor_store.items():
        plot_dashboard(sid, sd)
    consumer.close()
    print('Consumer closed.')
```
