```python
from kafka import KafkaProducer
import json
import time
import random
from math import cos, pi
```

```python
def generate_sensor_data():
    timestamp = int(time.time())

    # Simulate IoT sensor data for water quality metrics with realistic patterns
    water_temperature = random.uniform(1, 3) + round(20 + 5 * (1 + 0.5 * (1 + cos((timestamp % 86400) / 86400 * 2 * pi))), 2)
    ph_level = random.uniform(0, 1) + round(7.5 + 0.2 * (1 + cos((timestamp % 86400) / 86400 * 2 * pi)), 2)
    turbidity = round(random.uniform(5, 50), 2)  # Turbidity in NTU (Nephelometric Turbidity Units)
    dissolved_oxygen = round(random.uniform(5, 12), 2)  # Dissolved Oxygen in mg/L

    return {
        "timestamp": timestamp,
        "water_temperature": water_temperature,
        "ph_level": ph_level,
        "turbidity": turbidity,
        "dissolved_oxygen": dissolved_oxygen
    }
```

```python
 # Kafka configuration
kafka_topic = "water_quality"
kafka_bootstrap_servers = ["localhost:9092"] 

# Create Kafka producer
producer = KafkaProducer(
    bootstrap_servers=kafka_bootstrap_servers,
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

print(f"Producing messages to Kafka topic '{kafka_topic}'...")

try:
    while True:
        # Generate sensor data
        sensor_data = generate_sensor_data()

        # Publish sensor data to Kafka
        producer.send(kafka_topic, sensor_data)

        print(f"Sent: {sensor_data}")

        # Wait for 1 second
        time.sleep(1)
except KeyboardInterrupt:
    print("Stopped producing messages.")
finally:
    producer.close()
```

```python
import numpy as np
import time

def generate_sensor_data_np(n_samples=100): # n_samples define el tamaño del array
    timestamps = np.arange(int(time.time()), int(time.time()) + n_samples)
    
    phase = (timestamps % 86400) / 86400 * 2 * np.pi
    cos_phase = np.cos(phase)

    water_temp = np.random.uniform(1, 3, n_samples) + np.round(20 + 5 * (1 + 0.5 * (1 + cos_phase)), 2)
    ph_level = np.random.uniform(0, 1, n_samples) + np.round(7.5 + 0.2 * (1 + cos_phase), 2)
    turbidity = np.round(np.random.uniform(5, 50, n_samples), 2)
    dissolved_oxygen = np.round(np.random.uniform(5, 12, n_samples), 2)

    data = np.rec.fromarrays(
        [timestamps, water_temp, ph_level, turbidity, dissolved_oxygen],
        names='timestamp, water_temperature, ph_level, turbidity, dissolved_oxygen'
    )
    return data

sensor_data = generate_sensor_data_np(50)
print(sensor_data['water_temperature'])
```

```python
 # Kafka configuration
kafka_topic = "water_quality"
kafka_bootstrap_servers = ["localhost:9092"] 

# Create Kafka producer
producer = KafkaProducer(
    bootstrap_servers=kafka_bootstrap_servers,
    value_serializer=lambda v: json.dumps(v).encode('utf-8')
)

print(f"Producing messages to Kafka topic '{kafka_topic}'...")

try:
    while True:
        # Generate sensor data
        sensor_data = generate_sensor_data_np()

        # Publish sensor data to Kafka
        producer.send(kafka_topic, sensor_data)

        print(f"Sent: {sensor_data}")

        # Wait for 1 second
        time.sleep(1)
except KeyboardInterrupt:
    print("Stopped producing messages.")
finally:
    producer.close()
```

# EJERCICIO 2.6

Modifica el archivo water_data_producer_pr_2.ipynb para que se creen múltiples productores con un id de sensor único que se envíe como key del mensaje. Adjunta captura en la que se muestre el envío de mensajes de varios productores en el mismo notebook (1.25 punto).

```python
import json
import sys
import time
import random
from math import cos, pi
from kafka import KafkaProducer

kafka_topic = "water_quality_final"
kafka_bootstrap_servers = ["localhost:9092"]
producer_number = 10 

SENSOR_ID = sys.argv[1] if len(sys.argv) > 1 else "default_sensor"

def generate_sensor_data(sensor_id):
    timestamp = int(time.time())
    daily_cycle = (1 + cos((timestamp % 86400) / 86400 * 2 * pi))

    return {
        "sensor_id":         sensor_id,
        "timestamp":         timestamp,
        "water_temperature": round(random.uniform(1, 3) + 20 + 5 * (1 + 0.5 * daily_cycle), 2),
        "ph_level":          round(random.uniform(0, 1) + 7.5 + 0.2 * (1 + daily_cycle), 2),
        "turbidity":         round(random.uniform(5, 50), 2),
        "dissolved_oxygen":  round(random.uniform(5, 12), 2),
    }


producers   = []
producer_ids = []

for i in range(producer_number):
    producer_ids.append(f"sensor_{str(i).zfill(3)}")   
    producers.append(
        KafkaProducer(
            bootstrap_servers=kafka_bootstrap_servers,
            value_serializer=lambda v: json.dumps(v).encode("utf-8"),
        )
    )

print(f"Starting {producer_number} producers  |  CLI sensor id: {SENSOR_ID}")
print("─" * 60)


try:
    while True:                          
        for i in range(producer_number):
            sensor_data = generate_sensor_data(producer_ids[i])
            producers[i].send(
                kafka_topic,
                key=producer_ids[i].encode("utf-8"),  
                value=sensor_data,
            )

            print(f"Producer {producer_ids[i]} | Key: {producer_ids[i]} | {sensor_data}")

        time.sleep(2)

except KeyboardInterrupt:
    print("Stopped producing messages.")
    for p in producers:
        p.flush()
        p.close()

print("─" * 60)
print("All producers finished.")
```
